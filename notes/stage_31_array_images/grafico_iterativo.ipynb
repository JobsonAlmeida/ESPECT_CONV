```python
epoch_index = 0
time_window_index = 3

cube = power_array[
    epoch_index,
    :,
    :,
    :,
    time_window_index
]

n_frequencies = cube.shape[2]

print("Formato do cubo:", cube.shape)
print("Tipo de frequencies:", type(frequencies))
print("Formato de frequencies:", frequencies.shape)

first_slice = np.ma.masked_equal(
    cube[:, :, 0],
    0
)

color_map = plt.colormaps["hot"].copy()
color_map.set_bad(alpha=0)

figure, axis = plt.subplots(figsize=(7, 7))

image = axis.imshow(
    first_slice,
    cmap=color_map,
    origin="upper",
    interpolation="none",
    vmin=vmin,
    vmax=vmax
)

title = axis.set_title(
    f"Frequência: {frequencies[0]:.2f} Hz"
)


def update(frequency_index):
    frequency_slice = cube[:, :, frequency_index]

    masked_slice = np.ma.masked_equal(
        frequency_slice,
        0
    )

    image.set_data(masked_slice)

    title.set_text(
        f"Frequência: {frequencies[frequency_index]:.2f} Hz"
    )

    return image, title


animation = FuncAnimation(
    figure,
    update,
    frames=n_frequencies,
    interval=300,
    blit=False,
    repeat=True
)

plt.close(figure)

HTML(animation.to_jshtml())
```
